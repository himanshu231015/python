```python
## Boolean masking for Condition Based Selection.
s=pd.Series([1,2,3,4,5])
print(s[[True,False,True,False,True]])
print(s.loc[[True,False,True,False,True]])
print(s.iloc[[True,False,True,False,True]])
print(s.get([[True,False,True,False,True]]))
```

```python
## USing of collables in selection 
s=pd.Series([i for i in range(20)])
def odd_selection(s):
    return[True if i%2!=0 else False for i in range(s.size)]
    
print(s[odd_selection])
print(s)
```

```python
s=pd.Series([i for i in range(100)])
def odd_selection(s):
    return[True if i%7==0 else False for i in range(s.size)]
    
print(s[odd_selection])
```
