```python
import pandas as pd
pd.__version__
```

```python
lst = ['indore','mumbai','jaipur']
sr=pd.Series(lst)
print(type(sr))
print(sr)
```

```python
num=[1,2,3,4,5,6,7]
sri=pd.Series(num)
print(type(sri))
print(sri)
```

```python
dict={0:'tum',1:'mere',2:'rohit'}
sri1=pd.Series(dict)
print(type(sri1))
print(sri1)
```

```python
dict1={0:'tum','ram':'mere',2:20,20.3:20}
sri2=pd.Series(dict1)
print(type(sri2))
print(sri2)
```

```python
dict1={0:'tum','ram':'mere',2:20,20.3:20}
sri2=pd.Series(data=dict1,index=[0,'ram',2,20.3])
print(type(sri2))
print(sri2)
```

```python
list1=[1,2,3,4,5]
ser=pd.Series(list1,copy=False)
ser.iloc[0]=909

```

```python
import numpy as np
arr=np.array([1,2,3,4])
ser1=pd.Series(arr,copy=False)
ser1.iloc[0]=1919
print(ser1)
```

```python
dict1={0:'tum','ram':'mere',2:20,20.3:20}
sri2=pd.Series(data=dict1)
print(type(sri2))
print(sri2)
```

```python
name=['rohit','karan','yash']
sries=pd.Series(data=name,index=['r','k','y'])
print(sries)
```

```python
## dtype paramenters
num1=[10,30,20,50,40,0]
ser=pd.Series(data=num1,dtype=bool)
print(ser)
```

```python
num2=[10,30,20,50,40,0]
ser=pd.Series(data=num2,name='my number')
print(ser.name)
print(ser)
```
